### Definiciones

Sea $ \{X_1, \ldots, X_n\} $ v.a.i. cada una con $ f_{X_i}(x_i) $. El conjunto de v.a. $ X_1, \ldots, X_n $ se denomina una muestra aleatoria de tamaño $ n $. Cualquier función real de una muestra aleatoria $ T(X_1, \ldots, X_n) $ se llama **estadístico**. Sea $ X $ una v.a. con función de densidad $ f(x; \theta) $ que depende de los parámetros desconocidos $ \theta $.

Sea $ (X_1, \ldots, X_n) $ una m.a. de $ X $. Se tiene:

$$
f(x;\theta) = f(x_1, \ldots, x_n; \theta) = \prod_{i=1}^n f(x_i; \theta)
$$

donde $ x_i $ son los valores de los datos.

Un estimador de $ \theta $ es cualquier estadístico $ T(X_1, \ldots, X_n) $, y se denota por:

$$
\hat{\theta} = T(X_1, \ldots, X_n)
$$

> **Nota:** un estimador es una variable aleatoria.

### Tipos de estimadores:
- Puntual: en función de un solo valor
- Intervalo: en función de un rango de valores

### Inferencia sobre $ \mu $

Cuando no conocemos $ \sigma $, se estima con $ S = \sqrt{s^2} $

$$
\sqrt{n}\left(\frac{\bar{Y} - \mu}{S}\right) \sim t_{n-1}
$$

Si $ Z \sim \mathcal{N}(0,1) $ y $ W \sim \chi^2_n $, y $ Z $ y $ W $ son independientes:

$$
T = \frac{Z}{\sqrt{W/n}} = \sqrt{n}\left(\frac{\bar{Y} - \mu}{S}\right) \quad \text{(distribución t de Student)}
$$

### Distribución F

Si $ W_1 $ y $ W_2 $ son v.a.i. con $ W_1 \sim \chi^2_{n_1} $ y $ W_2 \sim \chi^2_{n_2} $:

$$
F = \frac{W_1 / n_1}{W_2 / n_2} = \frac{S_1^2 / \sigma_1^2}{S_2^2 / \sigma_2^2} \quad \text{(distribución F)}
$$

### Propiedades de los estimadores

Sea $ (X_1, \ldots, X_n) $ una m.a. que proviene de una v.a. con $ f_X(x) $, donde:

$$
\mathbb{E}(X) = \mu, \quad \text{Var}(X) = \sigma^2
$$

$$
T(X) = \bar{X} = \frac{X_1 + \cdots + X_n}{n}
$$

El estimador es **insesgado** si:

$$
\mathbb{E}(\bar{X}) = \mu
$$

Un estimador $ \hat{\theta} = T(X_1, \ldots, X_n) $ es **insesgado** de $ \theta $ si:

$$
\mathbb{E}[\hat{\theta}] = \theta
$$

Si $ \hat{\theta} $ es insesgado:

$$
\Rightarrow \text{su error cuadrático medio} = \mathbb{E}[(\hat{\theta} - \theta)^2] = \text{Var}(\hat{\theta})
$$

### Método de momentos

Si $ X $ es v.a. con esperanza $ \mu $ y $ n \in \mathbb{N} $:

$$
\mathbb{E}[X^n] \rightarrow \text{n-ésimo momento de } X
$$

$$
M_X^{(n)}(0) = \mathbb{E}[X^n] \quad \text{(momento poblacional)}
$$

$$
M_n = \frac{1}{n} \sum_{i=1}^n X_i^k \quad \text{(momento muestral)}
$$

**Método de momentos:** igualar momentos poblacionales con los muestrales.

---

**Ejemplo:**

Hallar los estimadores por el **método de momentos** para $ (X_1, \ldots, X_n) $, una m.a. de una población con densidad Gamma $ (\alpha, \beta) $.

Queremos encontrar estimadores para $ \alpha $ y $ \beta $.

Sabemos que si $ X_i \sim \text{Gamma}(\alpha, \beta) $:

$$
\mathbb{E}[X] = \frac{\alpha}{\beta}, \quad \text{Var}(X) = \frac{\alpha}{\beta^2}
$$

Igualamos momentos:

$$\begin{align*}
\mathbb{E}[X] &= \frac{1}{n} \sum x_i = \bar{x}
\\
\mathbb{E}[X^2] &= \frac{1}{n} \sum x_i^2
\\
\frac{\alpha}{\beta} = \bar{x} & \Rightarrow \beta = \frac{\alpha}{\bar{x}}
\end{align*}$$

Sustituyendo:

$$
\frac{\alpha}{\beta^2} = \frac{\alpha}{\left(\frac{\alpha}{\bar{x}}\right)^2} = \frac{\bar{x}^2}{\alpha} = \frac{1}{n} \sum x_i^2
$$

Despejando:

$$
\hat{\alpha} = \frac{n\bar{x}^2}{\sum (x_i - \bar{x})^2}, \quad \hat{\beta} = \frac{n \bar{x}}{\sum (x_i - \bar{x})^2}
$$

### Estimación por máxima verosimilitud para Bernoulli

Sea $ X_i \sim \text{Ber}(p) $, v.a.i.i.d.

$$\begin{align*}
\mathcal{L}(x_1, \ldots, x_n \mid p) &= \mathbb{P}(X_1 = x_1 \mid p) \cdots \mathbb{P}(X_n = x_n \mid p)
\\
&= p^{x_1}(1-p)^{1 - x_1} \cdots p^{x_n}(1 - p)^{1 - x_n} \\ &= p^{\sum x_i} (1 - p)^{n - \sum x_i}
\end{align*}$$

Log-verosimilitud:

$$
\ln \mathcal{L} = \sum x_i \ln(p) + (n - \sum x_i) \ln(1 - p)
$$

Derivamos e igualamos a cero:

$$\begin{align*}
\frac{d}{dp} \ln \mathcal{L} &= \frac{1}{p} \sum x_i - \frac{n - \sum x_i}{1 - p} = 0
\\
& \Rightarrow \frac{(1 - p)\sum x_i - p(n - \sum x_i)}{p(1 - p)} = 0 \\ & \Rightarrow p = \frac{1}{n} \sum x_i
\end{align*}$$

Por lo tanto, el estimador de máxima verosimilitud es:

$$
\hat{p} = \frac{\sum x_i}{n}
$$

Si $\hat{\theta} = \bar{X}$ es EMV de $\theta$

### Error cuadrático medio (ECM)

$$
\text{ECM}(\hat{\theta}) = \mathbb{E} \left[ (\hat{\theta} - \theta)^2 \right]
$$

Separando en componentes de **dispersión** y **sesgo**:

$$
\text{ECM}(\hat{\theta}) = \text{Var}(\hat{\theta}) + \left( \mathbb{E}[\hat{\theta}] - \theta \right)^2
$$

---

**Ejemplo**

Sea $X_1, \dots, X_n$ una muestra aleatoria de una variable $X \sim \text{Poisson}(\lambda)$.

**Objetivos:**

a) Encontrar el EMV  
b) Verificar si es insesgado  
c) Calcular el ECM  

Función de probabilidad:
$$
f(x_i ; \theta) = \frac{e^{-\theta} \theta^{x}}{x!}, \quad x = 0, 1, \dots
$$

### a) EMV para Poisson

$$ \begin{align*}
L(x_1, \ldots, x_n; \theta) &= \prod_{i=1}^{n} \frac{e^{-\theta} \theta^{x_i}}{x_i!} \\ &= \frac{e^{-n\theta} \theta^{\sum x_i}}{x_1! x_2! \dots x_n!}
\\
\ln L(x_1, \ldots, x_n; \theta) &= \ln \left( \frac{e^{-n\theta} \theta^{\sum x_i}}{x_1! \dots x_n!} \right)
\\ &= -n\theta + \sum x_i \ln(\theta) - \sum \ln(x_i!)
\\
\frac{\partial}{\partial \theta} \ln L(x_1, \ldots, x_n; \theta) &= -n + \sum x_i \cdot \frac{1}{\theta} = 0
\\
& \Rightarrow \frac{1}{\theta} \sum x_i = n \quad \\ & \Rightarrow \frac{1}{n} \sum x_i = \theta
\\
& \Rightarrow \hat{\theta} = \bar{x}
\end{align*}$$

---

### b) Insesgadez del estimador

$$ \begin{align*}
\mathbb{E}[\hat{\theta}] &= \mathbb{E} \left[ \frac{1}{n} \sum x_i \right] \\ &= \frac{1}{n} \sum \mathbb{E}[x_i] \\ &= \frac{1}{n} \cdot n \theta \\ &= \theta
\end{align*} $$

---

### c) Error Cuadrático Medio

$$ \begin{align*}
\text{ECM}(\hat{\theta}) \\ &= \text{Var}(\hat{\theta}) \\ &= \text{Var} \left( \frac{1}{n} \sum x_i \right) \\ &= \frac{1}{n^2} \text{Var} \left( \sum x_i \right)
\\
\\ &= \frac{1}{n^2} \cdot n \cdot \text{Var}(x_i) \\ &= \frac{n \theta}{n^2} = \frac{\theta}{n}
\\
\Rightarrow \text{ECM}(\hat{\theta}) &= \frac{\theta}{n}
\end{align*} $$

### Definición: Sesgo de un estimador

Sea $\hat{\theta}$ un estimador de $\theta$:

$$
B = \mathbb{E}[\hat{\theta}] - \theta
$$

1) Si $\mathbb{E}[\hat{\theta}] > \theta$, el estimador es **sesgado positivamente** (a)

2) Si $\mathbb{E}[\hat{\theta}] < \theta$, el estimador es **sesgado negativamente** (b)

Gráficamente:

- (a) $\theta < \mathbb{E}[\hat{\theta}]$  
- (b) $\theta > \mathbb{E}[\hat{\theta}]$

### Minimización del ECM

Si $X$ va con segundo momento de orden finito:

$$
g(u) = \mathbb{E}[(X - u)^2] \quad \text{(ECM)}
$$

Entonces, $g(u)$ se minimiza cuando $\mathbb{E}[X] = u$.

**Consecuencia:**

$$
\text{Var}(X) \leq \mathbb{E}[(X - u)^2]
$$

### Suficiencia de una estadística

Sea $X_1, \ldots, X_n$ una m.a. con densidad $f(\cdot;\theta)$, donde $\theta$ puede ser un vector.

Una estadística $T = t(X_1, \ldots, X_n)$ es una **estadística suficiente** ⟺  
la distribución de $X_1, \ldots, X_n$ dado $T = t$ **no depende de** $\theta$.

$$
\mathbb{P}(X_1 = t_1, X_2 = t_2, \ldots, X_n = t_n \mid T = t_i) = \text{no depende de } \theta
$$

---

### Teorema de Fisher-Neyman

Si $X_1, \ldots, X_n$ es una m.a. con función de densidad $f(x; \theta)$,  
y $T = t(X_1, \ldots, X_n)$ es una estadística cuya f.d.p. es $g(t; \theta)$,  
entonces **$T$ es suficiente para $\theta$** si y sólo si:

$$
L(x_1, \ldots, x_n; \theta) = H(x_1, \ldots, x_n) \cdot g(t(x_1, \ldots, x_n), \theta)
$$

donde $H(x_1, \ldots, x_n)$ **no depende de** $\theta$.

### Estimadores → UMVUE
(Estimadores Insesgados de Varianza Uniformemente Mínima)

Sea $X_1, \ldots, X_n$ una m.a. de $f(\cdot; \theta)$.  
Un estimador $T^* = t(X_1, \ldots, X_n)$ de $\tau(\theta)$ se define como un **UMVUE** de $\tau(\theta)$ si y sólo si:

1. $\mathbb{E}[T^*] = \tau(\theta)$ (insesgado)

2. $\text{Var}(T^*) \leq \text{Var}(T)$ para cualquier otro estimador insesgado $T = t(X_1, \ldots, X_n)$ de $\tau(\theta)$

---

### Desigualdad de Cramér–Rao

"Establece un límite inferior para la varianza de cualquier estimador"

$$
\text{Var}(T) \geq \frac{[\tau'(\theta)]^2}{n \, \mathbb{E} \left[ \left( \frac{\partial}{\partial \theta} \ln f(X_i; \theta) \right)^2 \right]}
$$

> Si $\text{Var}(T)$ alcanza el límite inferior:

$$
\Rightarrow \text{T es un estimador insesgado óptimo}
$$

### Estimador Máximo Verosímil

Existen distintos métodos para estimar parámetros desconocidos a partir de un conjunto de datos. El estimador máximo verosímil (MLE) responde a la siguiente pregunta:

¿Para qué valor de parámetro los datos observados tienen la probabilidad más alta?

**Ejemplo 1**. Una moneda se kanza 100 veces. Dado que se obtienen 55 soles, encontrar el MLE para la probabilidad $p$ de obtener sol en un solo lanzamiento.

Podemos pensar en contar el número de soles en los 100 lanzamientos, entonces la probabilidad de obtener 55 soles en este experimento, es la siguiente:

Si X: el número de soles que se obtiene al lanzar una moneda, se tiene que $X\sim Bin(100,p)$, es decir,

$$ P(\text{55 soles}; p) = \binom{100}{55}p^{55}(1-p)^{100-55}$$

**Función de verosímilitud**. Sea $X_1, X_2, \dots, X_n$ una muestra aleatoria con $f(x; \theta)$. La función de verosimilitud (likelihood) es:
    $$ f(x_1, \dots, x_n;\theta) = f(x_1;\theta)\cdots f(x_n;\theta)$$
    
Entonces,
$$ L(x_1, \dots, x_n;\theta) = \prod_{i=1}^n f(x_i;\theta)$$

Ejemplo. Si $X_1 \dots, X_n$ es una m.a. Poisson($\lambda$), etonces
$$ L(x_1, \dots, x_n;\theta) = \frac{e^{⁻n\lambda} \lambda ^{\sum x_i}}{\prod x_i !}$$

**Estimador Máximo Verosíimil**, es el valor de $\theta$, donde $L(x_1, \dots, x_n;\theta)$ alcanza el máximo.

Pasos:
    
- $$ L(x_1, \dots, x_n;\theta) = \prod_{i=1}^n f(x_i;\theta) \quad \text{(función de verosimilitud)}$$
- $$\ln L(x_1, \dots, x_n;\theta) $$
- $$ \frac{d\ln L(x_1, \dots, x_n;\theta)}{d\theta}$$
- $$ \frac{d\ln L(x_1, \dots, x_n;\theta)}{d\theta} = 0 $$
  y despejat $\theta$ para encontrar el máximo.
  
  Para nuestro caso, tenemos
$$ \frac{P(\text{55 soles}; p)}{dp} = \binom{100}{55}p^{55}(1-p)^{45}-45p^{55}(1-p)^{44} = 0$$
entonces, resolviendo para $p$, se tiene
$$ \binom{100}{55}p^{55}(1-p)^{45} = 45p^{55}(1-p)^{44}$$
entonces
$$55(1-p) = 45 p $$
entonces, $$p = \frac{100}{55}$$
Por lo tanto, el MLE para $p$, es $\hat{p} = 0.55$

---

**Ejemplo:**

```python
from scipy.stats import bernoulli, binom
import numpy as np
import sympy #para calculo simbolico y algebraico -> encontrar el estimador de manera analitica
from sympy.abc import x

#Definir a p como una variable simbolica positiva

p = sympy.symbols('p', positive = True)
f = p**55 * (1-p)**(100-55)
p_hat= sympy.solve(sympy.diff(f,p),p)[0]
print("El estimador de máxima verosimilitud es: ", p_hat)


#Otro ejemplo:
r = binom.rvs(1,0.7, size = 100)
r

f = p**x * (1-p)**(1-x)
J = np.prod([f.subs(x,i) for i in r])
logJ = sympy.expand_log(sympy.log(J)) #logaritmo
p_hat = sympy.solve(sympy.diff(logJ,p),p)[0]
print("El estimador de máxima verosimilitud es: ", p_hat)
```

$\color{crimson}{\text{Ejercicio 2.}}$ Sea una muestra aleatoria $X_1, \ldots, X_n \overset{iid}{\sim} \mathcal{N}(\mu, \sigma^2) $.
Queremos encontrar el estimador de máxima verosimilitud para $\mu$ y $\sigma^2$.

La función de densidad es:
$$ f(x; \mu, \sigma^2) = \frac{1}{\sqrt{2\pi \sigma^2}} \exp\left(-\frac{(x - \mu)^2}{2\sigma^2}\right) \cdot \mathbf{1}_{(-\infty, \infty)}(x)$$

Obtener el EMV para $\mu$ y $\sigma^2$

Por definición la función de verosimilitud
$$ \begin{align*}
L(\mu, \sigma^2) &= \prod_{i=1}^n f(x_i, \mu, \sigma^2)
\\ &= \prod_{i=1}^n \left(\frac{1}{\sqrt{2\pi}\sigma} \exp\left(-\frac{(x_i-\mu)^2}{2\sigma^2}\right)\right)
\end{align*}$$

Obtenemos su logaritmo:
$$ \begin{align*}
\log(L(\mu, \sigma^2)) &= \log\left[\prod_{i=1}^n \left(\frac{1}{\sqrt{2\pi\sigma^2}} \exp\left(-\frac{(x_i-\mu)^2}{2\sigma^2}\right)\right)\right]
\\ &= \sum_{i=1}^n \left(\log\left(\frac{1}{\sqrt{2\pi\sigma^2}}\right) - \frac{(x_i-\mu)^2}{2\sigma^2}\right)
\\ &= \sum_{i=1}^n \left(\log(2\pi\sigma^2)^{-1/2} - \frac{(x_i-\mu)^2}{2\sigma^2}\right)
\\ &= \sum_{i=1}^n \left(-\frac{1}{2}\log(2\pi\sigma^2) - \frac{(x_i-\mu)^2}{2\sigma^2}\right)
\\ &= \sum_{i=1}^n -\frac{1}{2}\log(2\pi\sigma^2) - \sum_{i=1}^n \frac{(x_i-\mu)^2}{2\sigma^2}
\\ * &= -\frac{n}{2}\log(2\pi\sigma^2) - \frac{1}{2\sigma^2}\sum_{i=1}^n (x_i-\mu)^2
\end{align*}$$

Ahora se debe de derivar a * respecto a $\mu$ para obtener el primer EMV:

$$ \begin{align*}
\frac{\partial \log(L(\mu,\sigma^2))}{\partial \mu} &= \frac{\partial}{\partial \mu}\left[-\frac{n}{2}\log(2\pi\sigma^2) - \frac{1}{2\sigma^2}\sum_{i=1}^n (x_i-\mu)^2\right]
\\ &= -\frac{\partial}{\partial \mu}\left(\frac{1}{2\sigma^2}\right)\sum_{i=1}^n (x_i-\mu)^2
\\ &= -\frac{1}{2\sigma^2}\sum_{i=1}^n 2(x_i-\mu)(-1)
\\ &= \frac{1}{\sigma^2}\sum_{i=1}^n (x_i-\mu)
\end{align*}$$

Igualamos a 0 y despejamos $\mu$:
$$\begin{align*}
\frac{1}{\sigma^2} &\sum_{i=1}^n (x_i - \mu) = 0 \\
&\sum_{i=1}^n (x_i - \mu) = 0 \\
&\sum_{i=1}^n x_i = \sum_{i=1}^n \mu \\
&\sum_{i=1}^n x_i = n\mu \\
\hat{\mu} = &\frac{\sum_{i=1}^n x_i}{n} = \bar{x}
\end{align*}$$

Ahora para encontrar el EMV de $\sigma^2$ usamos * y derivamos respecto a $\sigma^2$:

$$\begin{align*}
\frac{\partial}{\partial \sigma^2}\log(L(\mu,\sigma^2)) &= \frac{\partial}{\partial \sigma^2}\left[-\frac{n}{2}\log(2\pi\sigma^2) - \frac{1}{2\sigma^2}\sum_{i=1}^n (x_i-\mu)^2\right]
\\
&= -\frac{n}{2}\left(\frac{2\pi}{2\pi\sigma^2}\right) - \frac{\partial}{\partial \sigma^2}\left[\frac{1}{2\sigma^2} \sum_{i=1}^n (x_i-\mu)^2\right]
\\
&= -\frac{n}{2\sigma^2} + \frac{1}{\sigma^4} \sum_{i=1}^n (x_i-\mu)^2
\end{align*}$$

Igualamos a 0 y despejamos $\sigma^2$:

$$\begin{align*}
-\frac{n}{2\sigma^2} + \frac{1}{\sigma^4} \sum_{i=1}^n (x_i-\mu)^2 &= 0
\\
2\sigma^4\left(-\frac{n}{2\sigma^2} + \frac{1}{\sigma^4} \sum_{i=1}^n (x_i-\mu)^2\right) &= 2\sigma^4(0)
\\
-n\sigma^2 + 2 \sum_{i=1}^n(x_i-\mu)^2 &= 0
\\
n \sigma^2 &= 2 \sum_{i=1}^n (x_i-\mu)^2
\\
\hat{\sigma}^2 &= \frac{2 \sum_{i=1}^n (x_i-\mu)^2}{n}
\end{align*}$$


$\therefore$ Los EMV respectivamente son:

$$\hat{\mu} = \bar{x} \quad \wedge \quad \hat{\sigma}^2 = \frac{2}{n} \sum_{i=1}^n (x_i-\mu)^2$$

# $\color{crimson}{\text{Intervalos de confianza}}$

Un **intervalo de confianza (IC)** es un rango de valores que se utiliza para **estimar un parámetro poblacional desconocido** (como la media, proporción o coeficientes de un modelo). Este intervalo se basa en datos muestrales y nos da una idea de la **incertidumbre de la estimación**.

---

### Interpretación

Un intervalo de confianza del 95% significa que, **si se repitiera el muestreo muchas veces**, aproximadamente el 95% de los intervalos construidos **contendrían el valor verdadero del parámetro**.

> Importante: *No significa que haya un 95% de probabilidad de que el parámetro esté dentro del intervalo calculado en una muestra específica.*

---

### Cálculo Básico del IC para la Media

Cuando queremos estimar la **media poblacional** \(\mu\) a partir de una muestra, y si conocemos (o aproximamos) la desviación estándar, el IC del 95% se calcula como:

$$
IC = \bar{x} \pm z_{\frac{\alpha}{2}} \cdot \frac{s}{\sqrt{n}}
$$

Donde:

- $\bar{x}$ = media muestral  
- $s$ = desviación estándar muestral  
- $n$ = tamaño de la muestra  
- $z_{\alpha/2}$ = valor crítico de la distribución normal (1.96 para 95%)

---

**Ejemplo de Aplicación en Python**

Supongamos que tenemos una muestra de datos:


```python
import numpy as np
import scipy.stats as stats

# Simulamos una muestra
np.random.seed(0)
datos = np.random.normal(loc=50, scale=10, size=30)

# Parámetros muestrales
media_muestral = np.mean(datos)
desv_std = np.std(datos, ddof=1)
n = len(datos)
alfa = 0.05

# Error estándar
se = desv_std / np.sqrt(n)

# Valor crítico t (porque usamos desviación estimada, no poblacional)
t_critico = stats.t.ppf(1 - alfa/2, df=n-1)

# Intervalo de confianza
limite_inferior = media_muestral - t_critico * se
limite_superior = media_muestral + t_critico * se

print(f"Intervalo de confianza del 95%: [{limite_inferior:.2f}, {limite_superior:.2f}]")
```

---

### a) Intervalo de confianza para la media μ de una distribución normal con varianza conocida

Distribución: \( \mathcal{N}(\mu, \sigma^2) \)

$$
IC = \bar{x} \pm z_{\alpha/2} \cdot \frac{\sigma}{\sqrt{n}}
$$

---

### b) Intervalo de confianza para la media μ con varianza desconocida

- Muestras grandes (\( n > 30 \)):

$$
IC = \bar{x} \pm z_{\alpha/2} \cdot \frac{s}{\sqrt{n}}
$$

- Muestras pequeñas (\( n \leq 30 \)):

$$
IC = \bar{x} \pm t_{n-1; \alpha/2} \cdot \frac{s}{\sqrt{n}}
$$

---

### c) Intervalo de confianza para la varianza $ \sigma^2 $ de una distribución normal

$$
IC = \left[ \frac{(n-1)s^2}{\chi^2_{1 - \alpha/2, \, n-1}}, \; \frac{(n-1)s^2}{\chi^2_{\alpha/2, \, n-1}} \right]
$$

---

### d) Intervalo de confianza para la diferencia de medias de dos poblaciones normales

- Varianzas conocidas:

$$
IC = (\bar{x}_1 - \bar{x}_2) \pm z_{\alpha/2} \cdot \sqrt{\frac{\sigma_1^2}{n_1} + \frac{\sigma_2^2}{n_2}}
$$

- Varianzas desconocidas, muestras grandes (\( n_1 + n_2 > 30 \), \( n_1 \approx n_2 \)):

$$
IC = (\bar{x}_1 - \bar{x}_2) \pm z_{\alpha/2} \cdot \sqrt{\frac{s_1^2}{n_1} + \frac{s_2^2}{n_2}}
$$

- Muestras pequeñas, varianzas desconocidas pero iguales:

$$\begin{align*}
IC &= (\bar{x}_1 - \bar{x}_2) \pm t_{n_1 + n_2 - 2; \alpha/2} \cdot \sqrt{s_p^2 \left( \frac{1}{n_1} + \frac{1}{n_2} \right)}
\\
s_p^2 &= \frac{(n_1 - 1)s_1^2 + (n_2 - 1)s_2^2}{n_1 + n_2 - 2}
\end{align*}$$

- Muestras pequeñas, varianzas desconocidas y distintas:

$$\begin{align*}
IC &= (\bar{x}_1 - \bar{x}_2) \pm t_{f; \alpha/2} \cdot \sqrt{ \frac{s_1^2}{n_1} + \frac{s_2^2}{n_2} }
\\
f &= \frac{ \left( \frac{s_1^2}{n_1} + \frac{s_2^2}{n_2} \right)^2 }{ \frac{(s_1^2/n_1)^2}{n_1 - 1} + \frac{(s_2^2/n_2)^2}{n_2 - 1} } \quad \text{(Aproximación de Welch)}
\end{align*}$$

---

### e) Intervalo de confianza para la razón de varianzas

$$
IC = \left[ \frac{s_1^2}{s_2^2} \cdot \frac{1}{F_{1 - \alpha/2; n_1-1, n_2-1}}, \; \frac{s_1^2}{s_2^2} \cdot F_{\alpha/2; n_1-1, n_2-1} \right]
$$

---

### f) Intervalo de confianza para la proporción $ p $ de una distribución binomial $ B(n, p) $

$$
IC = \hat{p} \pm z_{\alpha/2} \cdot \sqrt{ \frac{\hat{p}(1 - \hat{p})}{n} }
$$

---

### g) Intervalo de confianza para la diferencia de proporciones $ p_1 - p_2 $

$$
IC = (\hat{p}_1 - \hat{p}_2) \pm z_{\alpha/2} \cdot \sqrt{ \frac{\hat{p}_1 (1 - \hat{p}_1)}{n_1} + \frac{\hat{p}_2 (1 - \hat{p}_2)}{n_2} }
$$

---

### h) Intervalo de confianza para el parámetro \( \lambda \) de una distribución de Poisson

$$
IC = \lambda \pm z_{\alpha/2} \cdot \sqrt{ \frac{\lambda}{n} }
$$

---

### i) Intervalo de confianza para datos apareados

- Muestras grandes (\( n > 30 \)):

$$
IC = \bar{d} \pm z_{\alpha/2} \cdot \frac{s_d}{\sqrt{n}}
$$

- Muestras pequeñas (\( n \leq 30 \)):

$$
IC = \bar{d} \pm t_{n-1; \alpha/2} \cdot \frac{s_d}{\sqrt{n}}
$$

Donde:

$$
\bar{d} = \frac{1}{n} \sum_{i=1}^{n} d_i \quad \text{y} \quad d_i = X_i - Y_i
$$

$$
s_d^2 = \frac{1}{n - 1} \sum_{i=1}^{n} (d_i - \bar{d})^2
$$

$\color{red}{\text{Ejercicio 1:}}$ En una encuesta de una Universidad, 225 estudiantes son seleccionados al azar de los que toman cálculo y se les pregunta si cálculo es su materia favorita. 100 estudiantes responden que el cálculo es su materia favorita. Proporcione un intervalo de confianza del 95 %  para la proporción de todos los estudiantes que toman cálculo y que la consideran su materia favorita.  

```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as st
from scipy.stats import norm

# Datos del problema
n = 225  # Tamaño de la muestra
x = 100  # Número de éxitos (estudiantes que consideran cálculo su materia favorita)
alpha = 0.05  # Nivel de significancia para un intervalo de confianza del 95%

# Calcular la proporción muestral
p_hat = x / n

# Calcular el error estándar
error_estandar = np.sqrt((p_hat * (1 - p_hat)) / n)

# Valor crítico z para un nivel de confianza del 95% (alpha = 0.05)
z_alpha_2 = st.norm.ppf(1 - alpha/2)

# Calcular el margen de error
margen_error = z_alpha_2 * error_estandar

# Calcular los límites del intervalo de confianza
limite_inferior = p_hat - margen_error
limite_superior = p_hat + margen_error

print(f"\nIntervalo de confianza del 95% para la proporción: ({limite_inferior:.4f}, {limite_superior:.4f})")
print(f"\nInterpretación: Con un 95% de confianza, la proporción de estudiantes que consideran cálculo su materia favorita está entre {limite_inferior:.4f} y {limite_superior:.4f}")

# Formato en porcentaje para una interpretación más intuitiva
print(f"\nEn porcentaje: ({limite_inferior*100:.2f}%, {limite_superior*100:.2f}%)")
```

$\color{red}{\text{Ejercicio 2:}}$ Los datos $1.2,\ 2.1,\ 2.3,\ 1.9,\ 3.0,\ 1.5,\ 2.8,\ 2.4,\ 1.7,\ 2.6,\ 2.0,\ 1.8,\ 3.1,\ 2.2,\ 2.5$ se toman a partir de una distribución normal $N(\mu,\sigma^2)$ con $\mu$ desconocida.

**a)** Encuentra un intervalo de confianza al nivel de $90\%$ para $\mu$, dado que $\sigma =2$.

**b)** Encuentra un intervalo de confianza al nivel de $90\%$ para $\mu$.

**c)** Encuentra un intervalo de confianza al nivel de $90\%$ para $\sigma^2$.

```python
# Datos proporcionados
datos = np.array([1.2, 2.1, 2.3, 1.9, 3.0, 1.5, 2.8, 2.4, 1.7, 2.6, 2.0, 1.8, 3.1, 2.2, 2.5])

# Calcular estadísticas básicas
n = len(datos)
media_muestral = np.mean(datos)
desv_est_muestral = np.std(datos, ddof=1)  # ddof=1 para obtener la desviación estándar muestral
varianza_muestral = desv_est_muestral**2

# Nivel de confianza
nivel_confianza = 0.90
alpha = 1 - nivel_confianza

# a) Intervalo de confianza para μ con σ conocida (σ=2)
sigma_conocida = 2
z_alpha_2 = st.norm.ppf(1 - alpha/2)
error_estandar_a = sigma_conocida / np.sqrt(n)
margen_error_a = z_alpha_2 * error_estandar_a

limite_inferior_a = media_muestral - margen_error_a
limite_superior_a = media_muestral + margen_error_a

print(f"\na) Intervalo de confianza al 90% para μ con σ=2: ({limite_inferior_a:.4f}, {limite_superior_a:.4f})")

# b) Intervalo de confianza para μ con σ desconocida
# Usamos la distribución t de Student
t_alpha_2 = st.t.ppf(1 - alpha/2, df=n-1)
error_estandar_b = desv_est_muestral / np.sqrt(n)
margen_error_b = t_alpha_2 * error_estandar_b

limite_inferior_b = media_muestral - margen_error_b
limite_superior_b = media_muestral + margen_error_b

print(f"\nb) Intervalo de confianza al 90% para μ con σ desconocida: ({limite_inferior_b:.4f}, {limite_superior_b:.4f})")

# c) Intervalo de confianza para σ² (varianza)
# Para la varianza usamos la distribución chi-cuadrado
chi2_lower = st.chi2.ppf(alpha/2, df=n-1)
chi2_upper = st.chi2.ppf(1 - alpha/2, df=n-1)

limite_inferior_c = (n-1) * varianza_muestral / chi2_upper
limite_superior_c = (n-1) * varianza_muestral / chi2_lower

print(f"\nc) Intervalo de confianza al 90% para σ²: ({limite_inferior_c:.4f}, {limite_superior_c:.4f})")

# También podemos mostrar el intervalo para σ (desviación estándar)
print(f"   Intervalo de confianza al 90% para σ: ({np.sqrt(limite_inferior_c):.4f}, {np.sqrt(limite_superior_c):.4f})")
```

Suponga que este promedio de deuda se basa en una muestra de 480 préstamos a estudiantes y que la desviación estándar de la población de las deudas al recibirse es \$2,000.

**a)** Determina un estimado de confianza del **90%** del promedio poblacional de la deuda.

**b)** Determina un estimado de confianza del **95%** del promedio poblacional de la deuda.

**c)** Determina un estimado de confianza del **99%** del promedio poblacional de la deuda.

**d)** Describe lo que sucede con el **ancho del intervalo de confianza** a medida que se **aumenta el nivel de confianza**. ¿Parece razonable? Explica tu respuesta.

```python
# Datos
media_muestral = 12658
n = 480
sigma = 2000

# Error estándar
se = sigma / np.sqrt(n)

# Niveles de confianza y sus colores
niveles_confianza = {
    "90%": {"confianza": 0.90, "color": "skyblue"},
    "95%": {"confianza": 0.95, "color": "lightgreen"},
    "99%": {"confianza": 0.99, "color": "salmon"}
}

# Crear la figura
fig, ax = plt.subplots(figsize=(10, 6))

# Rango para la distribución normal
x = np.linspace(media_muestral - 4 * se, media_muestral + 4 * se, 1000)
y = norm.pdf(x, media_muestral, se)

# Graficar la distribución
ax.plot(x, y, label='Distribución de la media muestral', color='black')

# Agregar los intervalos de confianza
for i, (nivel, datos) in enumerate(niveles_confianza.items(), start=1):
    z = norm.ppf(1 - (1 - datos["confianza"]) / 2)
    margen_error = z * se
    li = media_muestral - margen_error
    ls = media_muestral + margen_error
    print(f"Intervalo de confianza del {nivel}: [{li:.2f}, {ls:.2f}]")

    # Sombrear el intervalo de confianza
    x_fill = np.linspace(li, ls, 100)
    y_fill = norm.pdf(x_fill, media_muestral, se)
    ax.fill_between(x_fill, y_fill, alpha=0.5, label=f'{nivel} CI', color=datos["color"])

# Añadir etiquetas
ax.axvline(media_muestral, color='red', linestyle='--', label='Media muestral')
ax.set_title('Intervalos de confianza sobre la media muestral')
ax.set_xlabel('Deuda en dólares')
ax.set_ylabel('Densidad de probabilidad')
ax.legend()
ax.grid(True)

plt.tight_layout()
plt.show()
```

**d)**

A medida que aumenta el nivel de confianza, también aumenta el valor de $Z$ y por lo tanto el ancho del intervalo de confianza se vuelve mayor.

- $90\% \rightarrow \text{intervalo más estrecho}$

- $95\% \rightarrow \text{más ancho}$

- $99\% \rightarrow \text{aún más ancho}$

**¿Parece razonable?** Sí, porque si queremos estar más seguros de que el intervalo contiene el verdadero valor de la media poblacional, necesitamos cubrir una mayor parte de la distribución, lo cual se refleja en un intervalo más amplio.

# $\color{crimson}{\text{ Pruebas de Hipótesis}}$

Las pruebas de hipótesis son procedimientos estadísticos que nos permiten tomar decisiones sobre una población basándonos en información muestral. Básicamente, evaluamos afirmaciones sobre los parámetros de una población.

### Hipótesis Estadísticas

- **Hipótesis nula ($H_0$)**: Es la afirmación que asumimos como verdadera inicialmente y buscamos evidencia para rechazarla. Generalmente establece que "no hay efecto" o "no hay diferencia".
- **Hipótesis alternativa ($H_1$ o $H_a$)**: Es la afirmación contraria a la hipótesis nula. Representa lo que estamos tratando de demostrar.

### Tipos de Pruebas

- **Prueba bilateral (dos colas)**: $H_1: \theta \neq \theta_0$
- **Prueba unilateral derecha (cola derecha)**: $H_1: \theta > \theta_0$
- **Prueba unilateral izquierda (cola izquierda)**: $H_1: \theta < \theta_0$

Donde $\theta$ es el parámetro poblacional y $\theta_0$ es un valor específico.

### Proceso de una Prueba de Hipótesis

1. **Establecer las hipótesis** ($H_0$ y $H_1$)
2. **Seleccionar el nivel de significancia** ($\alpha$)
3. **Calcular el estadístico de prueba**
4. **Determinar el valor p o la región crítica**
5. **Tomar una decisión estadística**
6. **Interpretar la decisión en el contexto del problema**

### Errores en la Prueba de Hipótesis

| | $H_0$ es verdadera | $H_0$ es falsa |
|---|---|---|
| No rechazar $H_0$ | Decisión correcta | Error Tipo II (β) |
| Rechazar $H_0$ | Error Tipo I (α) | Decisión correcta |

- **Error Tipo I**: Rechazar $H_0$ cuando es verdadera (falso positivo)
- **Error Tipo II**: No rechazar $H_0$ cuando es falsa (falso negativo)

### Implementación en Python

```python
# Pruebas de hipótesis con Python
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt
import seaborn as sns

# Configuración visual
sns.set_style('darkgrid')  # Usamos un estilo válido de seaborn
sns.set_context('notebook')

# Ejemplo 1: Prueba t para una muestra
# Supongamos que queremos probar si la media de una población es igual a 10

# Generamos datos simulados
np.random.seed(42)
muestra = np.random.normal(loc=10.5, scale=2, size=50)

# Realizamos la prueba t
t_stat, p_valor = stats.ttest_1samp(muestra, 10)

print(f"Estadístico t: {t_stat:.4f}")
print(f"Valor p: {p_valor:.4f}")

# Decisión estadística (con alfa = 0.05)
if p_valor < 0.05:
    print("Rechazamos H0")
else:
    print("No rechazamos H0")

# Visualización
plt.figure(figsize=(10, 6))
plt.hist(muestra, bins=15, alpha=0.7, color='skyblue', edgecolor='black')
plt.axvline(x=10, color='red', linestyle='--', label='Valor de H0 (μ=10)')
plt.axvline(x=np.mean(muestra), color='green', linestyle='-', label=f'Media muestral ({np.mean(muestra):.2f})')
plt.title('Distribución de la Muestra vs Hipótesis Nula')
plt.xlabel('Valores')
plt.ylabel('Frecuencia')
plt.legend()
plt.show()
```

```python
# Prueba Z para una Proporción

# Supongamos que queremos probar si la proporción de éxitos es igual a 0.5
# H0: p = 0.5
# H1: p ≠ 0.5

# Datos
n = 100  # tamaño de muestra
x = 60   # número de éxitos
p0 = 0.5  # proporción bajo la hipótesis nula

# Calcular estadístico z y valor p
p_muestra = x / n
z = (p_muestra - p0) / np.sqrt(p0 * (1 - p0) / n)
p_valor = 2 * (1 - stats.norm.cdf(abs(z)))  # prueba bilateral

print(f"Proporción muestral: {p_muestra:.2f}")
print(f"Estadístico z: {z:.4f}")
print(f"Valor p: {p_valor:.4f}")

# Decisión estadística (con alfa = 0.05)
if p_valor < 0.05:
    print("Rechazamos H0")
else:
    print("No rechazamos H0")
```

### Principales Pruebas de Hipótesis

| Prueba | Uso |
|---|---|
| Prueba Z | Para medias o proporciones con muestras grandes |
| Prueba t | Para medias con muestras pequeñas o varianza desconocida |
| ANOVA | Para comparar medias de más de dos grupos |
| Chi-cuadrado | Para independencia o bondad de ajuste |
| Prueba F | Para comparar varianzas |
| Mann-Whitney U | Alternativa no paramétrica a la prueba t independiente |
| Wilcoxon | Alternativa no paramétrica a la prueba t pareada |

### Potencia de la Prueba

La potencia de una prueba estadística es la probabilidad de rechazar correctamente la hipótesis nula cuando es falsa. Está relacionada con:

- Tamaño de la muestra
- Magnitud del efecto
- Nivel de significancia
- Variabilidad de los datos


```python
# Ejemplo de cálculo de potencia para prueba t
from statsmodels.stats.power import TTestPower

# Parámetros
effect_size = 0.5  # Tamaño del efecto (d de Cohen)
alpha = 0.05       # Nivel de significancia
n = 50             # Tamaño de muestra

# Calcular potencia
power_analysis = TTestPower()
power = power_analysis.power(effect_size=effect_size, nobs=n, alpha=alpha)

print(f"Potencia de la prueba: {power:.4f}")

# Gráfico de potencia vs tamaño de muestra
sample_sizes = np.arange(10, 100, 5)
power_values = [power_analysis.power(effect_size=effect_size, nobs=n, alpha=alpha) for n in sample_sizes]

plt.figure(figsize=(10, 6))
plt.plot(sample_sizes, power_values, 'o-', color='blue')
plt.axhline(y=0.8, color='red', linestyle='--', label='Potencia deseada (0.8)')
plt.title('Potencia vs Tamaño de Muestra')
plt.xlabel('Tamaño de Muestra')
plt.ylabel('Potencia')
plt.grid(True)
plt.legend()
plt.show()
```

### Valor P vs Nivel de Significancia

- **Valor p**: Probabilidad de obtener un resultado al menos tan extremo como el observado, dado que la hipótesis nula es verdadera.
- **Nivel de significancia (α)**: Umbral predefinido para rechazar H0 (típicamente 0.05 o 0.01).

Regla de decisión:
- Si p-valor < α, rechazamos H0
- Si p-valor ≥ α, no rechazamos H0

### $\color{lightcoral}{\text{CONTRASTE DE HIPÓTESIS}}$

### a) Contraste de la media de una población normal con varianza conocida

**Contraste bilateral**
- Hipótesis nula $H_0: \mu = \mu_0$
- Hipótesis alternativa $H_a: \mu \neq \mu_0$

Se acepta $H_0$ si el estadístico $z = \frac{x - \mu_0}{\sigma/\sqrt{n}} \leq z_{\alpha/2}$

Se rechaza $H_0$ si el estadístico $z = \frac{x - \mu_0}{\sigma/\sqrt{n}} > z_{\alpha/2}$

Región de rechazo: $R_{rechazo} = \{x | x - \mu_0 > z_{\alpha/2} \cdot \frac{\sigma}{\sqrt{n}}\}$

**Contraste unilateral**
- Hipótesis nula $H_0: \mu \leq \mu_0$
- Hipótesis alternativa $H_a: \mu > \mu_0$

Se acepta $H_0$ si el estadístico $z = \frac{x - \mu_0}{\sigma/\sqrt{n}} \leq z_{\alpha}$

Se rechaza $H_0$ si el estadístico $z = \frac{x - \mu_0}{\sigma/\sqrt{n}} > z_{\alpha}$

Región de rechazo: $R_{rechazo} = \{x | x - \mu_0 > z_{\alpha} \cdot \frac{\sigma}{\sqrt{n}}\}$

### b) Contraste de la media de una población normal con varianza desconocida

### Contraste bilateral

**Muestras grandes $n > 30$**
- Hipótesis nula $H_0: \mu = \mu_0$
- Hipótesis alternativa $H_a: \mu \neq \mu_0$

Se acepta $H_0$ si el estadístico $z = \frac{x - \mu_0}{s/\sqrt{n}} \leq z_{\alpha/2}$

Se rechaza $H_0$ si el estadístico $z = \frac{x - \mu_0}{s/\sqrt{n}} > z_{\alpha/2}$

Región de rechazo: $R_{rechazo} = \{x | x - \mu_0 > z_{\alpha/2} \cdot \frac{s}{\sqrt{n}}\}$

**Muestras pequeñas $n \leq 30$**
- Hipótesis nula $H_0: \mu = \mu_0$
- Hipótesis alternativa $H_a: \mu \neq \mu_0$

Se acepta $H_0$ si el estadístico $t = \frac{x - \mu_0}{s/\sqrt{n}} \leq t_{\alpha/2;n-1}$

Se rechaza $H_0$ si el estadístico $t = \frac{x - \mu_0}{s/\sqrt{n}} > t_{\alpha/2;n-1}$

Región de rechazo: $R_{rechazo} = \{x | x - \mu_0 > t_{\alpha/2;n-1} \cdot \frac{s}{\sqrt{n}}\}$

**Muestras grandes $n > 30$**

Se rechaza $H_0$ si el estadístico $z = \frac{x - \mu_0}{s/\sqrt{n}} > z_{\alpha}$

Región de rechazo: $R_{rechazo} = \{x | x - \mu_0 > z_{\alpha} \cdot \frac{s}{\sqrt{n}}\}$

**Muestras pequeñas $n \leq 30$**
- Hipótesis nula $H_0: \mu \leq \mu_0$
- Hipótesis alternativa $H_a: \mu > \mu_0$

Se acepta $H_0$ si el estadístico $t = \frac{x - \mu_0}{s/\sqrt{n}} \leq t_{\alpha;n-1}$

### c) Contraste para la varianza de una población normal

**Contraste bilateral**
- Hipótesis nula $H_0: \sigma^2 = \sigma_0^2$
- Hipótesis alternativa $H_a: \sigma^2 \neq \sigma_0^2$

Se acepta $H_0$ si el estadístico $\chi^2 = \frac{(n-1)s^2}{\sigma_0^2} \in [\chi^2_{\alpha/2;(n-1)}, \chi^2_{1-\alpha/2;(n-1)}]$

Se rechaza $H_0$ si el estadístico $\chi^2 = \frac{(n-1)s^2}{\sigma_0^2} \notin [\chi^2_{\alpha/2;(n-1)}, \chi^2_{1-\alpha/2;(n-1)}]$

**Contraste unilateral**
- Hipótesis nula $H_0: \sigma^2 \leq \sigma_0^2$
- Hipótesis alternativa $H_a: \sigma^2 > \sigma_0^2$

Se acepta $H_0$ si el estadístico $\chi^2 = \frac{(n-1)s^2}{\sigma_0^2} < \chi^2_{\alpha;(n-1)}$

Se rechaza $H_0$ si el estadístico $\chi^2 = \frac{(n-1)s^2}{\sigma_0^2} \geq \chi^2_{\alpha;(n-1)}$

### d) Contraste de igualdad de medias de dos poblaciones normales de varianzas conocidas

**Contraste bilateral**
- Hipótesis nula $H_0: \mu_1 = \mu_2$
- Hipótesis alternativa $H_a: \mu_1 \neq \mu_2$

Se acepta $H_0$ si $z = \frac{\bar{x} - \bar{y}}{\sqrt{\frac{\sigma_1^2}{n_1} + \frac{\sigma_2^2}{n_2}}} \leq z_{\alpha/2}$

Se rechaza $H_0$ si $z = \frac{\bar{x} - \bar{y}}{\sqrt{\frac{\sigma_1^2}{n_1} + \frac{\sigma_2^2}{n_2}}} > z_{\alpha/2}$

**Contraste unilateral**
- Hipótesis nula $H_0: \mu_1 \leq \mu_2$
- Hipótesis alternativa $H_a: \mu_1 > \mu_2$

Se rechaza $H_0$ si $z = \frac{\bar{x} - \bar{y}}{\sqrt{\frac{\sigma_1^2}{n_1} + \frac{\sigma_2^2}{n_2}}} > z_{\alpha}$

### e) Contraste de igualdad de medias de dos poblaciones normales de varianzas desconocidas

### Contraste bilateral
- Hipótesis nula $H_0: \mu_1 = \mu_2$
- Hipótesis alternativa $H_a: \mu_1 \neq \mu_2$

#### Muestras grandes $(n_1 + n_2) > 30$ con $n_1 \approx n_2$

Se acepta $H_0$ si el estadístico $z = \frac{\bar{x} - \bar{y}}{\sqrt{\frac{s_1^2}{n_1} + \frac{s_2^2}{n_2}}} \leq z_{\alpha/2}$

Se rechaza $H_0$ si el estadístico $z = \frac{\bar{x} - \bar{y}}{\sqrt{\frac{s_1^2}{n_1} + \frac{s_2^2}{n_2}}} > z_{\alpha/2}$

#### Muestras pequeñas $(n_1 + n_2) \leq 30$, varianzas desconocidas pero iguales $\sigma_1^2 = \sigma_2^2$

Se acepta $H_0$ si $t = \frac{\bar{x} - \bar{y}}{s_p\sqrt{\frac{1}{n_1} + \frac{1}{n_2}}} \leq t_{\alpha/2;n_1+n_2-2}$

donde $s_p^2 = \frac{(n_1-1)s_1^2 + (n_2-1)s_2^2}{n_1+n_2-2}$

Se rechaza $H_0$ si el estadístico $t = \frac{\bar{x} - \bar{y}}{s_p\sqrt{\frac{1}{n_1} + \frac{1}{n_2}}} > t_{\alpha/2;n_1+n_2-2}$

#### Muestras pequeñas $(n_1 + n_2) \leq 30$, varianzas desconocidas y distintas $\sigma_1^2 \neq \sigma_2^2$

Se acepta $H_0$ si $t = \frac{\bar{x} - \bar{y}}{\sqrt{\frac{s_1^2}{n_1} + \frac{s_2^2}{n_2}}} \leq t_{\alpha/2;f}$

donde $f = \frac{\left(\frac{s_1^2}{n_1} + \frac{s_2^2}{n_2}\right)^2}{\frac{(s_1^2/n_1)^2}{n_1-1} + \frac{(s_2^2/n_2)^2}{n_2-1}}$

Se rechaza $H_0$ si el estadístico $t = \frac{\bar{x} - \bar{y}}{\sqrt{\frac{s_1^2}{n_1} + \frac{s_2^2}{n_2}}} > t_{\alpha/2;f}$

#### Muestras pequeñas $(n_1 + n_2) \leq 30$, varianzas desconocidas y distintas $\sigma_1^2 \neq \sigma_2^2$

Se acepta $H_0$ si $t = \frac{\bar{x} - \bar{y}}{\sqrt{\frac{s_1^2}{n_1} + \frac{s_2^2}{n_2}}} \leq t_{\alpha;f}$

donde $f = \frac{\left(\frac{s_1^2}{n_1} + \frac{s_2^2}{n_2}\right)^2}{\frac{(s_1^2/n_1)^2}{n_1-1} + \frac{(s_2^2/n_2)^2}{n_2-1}}$

Se rechaza $H_0$ si el estadístico $t = \frac{\bar{x} - \bar{y}}{\sqrt{\frac{s_1^2}{n_1} + \frac{s_2^2}{n_2}}} > t_{\alpha;f}$

### f) Contraste de igualdad de varianzas de dos poblaciones normales

**Contraste bilateral**
- Hipótesis nula $H_0: \sigma_1^2 = \sigma_2^2$
- Hipótesis alternativa $H_a: \sigma_1^2 \neq \sigma_2^2$

Se acepta $H_0$ si el estadístico $F = \frac{s_1^2}{s_2^2} \in [F_{1-\alpha/2;(n_1-1),(n_2-1)};F_{\alpha/2;(n_1-1),(n_2-1)}]$

Se rechaza $H_0$ si el estadístico $F = \frac{s_1^2}{s_2^2} \notin [F_{1-\alpha/2;(n_1-1),(n_2-1)};F_{\alpha/2;(n_1-1),(n_2-1)}]$

**Contraste unilateral**
- Hipótesis nula $H_0: \sigma_1^2 \leq \sigma_2^2$
- Hipótesis alternativa $H_a: \sigma_1^2 > \sigma_2^2$

Se acepta $H_0$ si $F = \frac{s_1^2}{s_2^2} \leq F_{\alpha;(n_1-1),(n_2-1)}$

Se rechaza $H_0$ si $F = \frac{s_1^2}{s_2^2} > F_{\alpha;(n_1-1),(n_2-1)}$

> **NOTA:** $F_{\alpha;n_1,n_2} = \frac{1}{F_{1-\alpha;n_2,n_1}}$

### g) Contraste de igualdad de medias en el caso de datos apareados

**Contraste bilateral**
- Hipótesis nula $H_0: \mu_1 = \mu_2 \Leftrightarrow d = 0$
- Hipótesis alternativa $H_a: \mu_1 \neq \mu_2 \Leftrightarrow d \neq 0$

Se acepta $H_0$ si el estadístico $t = \frac{\bar{d}}{s_d/\sqrt{n}} \leq t_{\alpha/2,n-1}$

Se rechaza $H_0$ si el estadístico $t = \frac{\bar{d}}{s_d/\sqrt{n}} > t_{\alpha/2,n-1}$

donde $\bar{d} = \frac{1}{n}\sum_{i=1}^{n}d_i = \frac{1}{n}\sum_{i=1}^{n}(x_i-y_i)$ y $s_d^2 = \frac{1}{n-1}\sum_{i=1}^{n}(d_i-\bar{d})^2$

**Contraste unilateral**
- Hipótesis nula $H_0: \mu_1 \leq \mu_2 \Leftrightarrow d \leq 0$
- Hipótesis alternativa $H_a: \mu_1 > \mu_2 \Leftrightarrow d > 0$

Se acepta $H_0$ si el estadístico $t = \frac{\bar{d}}{s_d/\sqrt{n}} \leq t_{\alpha,n-1}$

Se rechaza $H_0$ si el estadístico $t = \frac{\bar{d}}{s_d/\sqrt{n}} > t_{\alpha,n-1}$

### h) Contraste para el parámetro p de una distribución binomial

**Contraste bilateral**
- Hipótesis nula $H_0: p = p_o$
- Hipótesis alternativa $H_a: p \neq p_o$

Se acepta $H_0$ si $z = \frac{p - p_o}{\sqrt{\frac{p_o(1-p_o)}{n}}} \leq z_{\alpha/2}$

Se rechaza $H_0$ si $z = \frac{p - p_o}{\sqrt{\frac{p_o(1-p_o)}{n}}} > z_{\alpha/2}$

**Contraste unilateral**
- Hipótesis nula $H_0: p \leq p_o$
- Hipótesis alternativa $H_a: p > p_o$

Se acepta $H_0$ si $z = \frac{p - p_o}{\sqrt{\frac{p_o(1-p_o)}{n}}} \leq z_{\alpha}$

Se rechaza $H_0$ si $z = \frac{p - p_o}{\sqrt{\frac{p_o(1-p_o)}{n}}} > z_{\alpha}$

### i) Contraste para la igualdad de los parámetros de dos distribuciones binomiales $B(n_1, p_1)$ y $B(n_2, p_2)$ en muestras grandes

**Contraste bilateral**
- Hipótesis nula $H_0: p_1 = p_2$
- Hipótesis alternativa $H_a: p_1 \neq p_2$

Se acepta $H_0$ si el estadístico $z = \frac{p_1 - p_2}{\sqrt{\frac{p_1(1-p_1)}{n_1} + \frac{p_2(1-p_2)}{n_2}}} \leq z_{\alpha/2}$

Se rechaza $H_0$ si el estadístico $z = \frac{p_1 - p_2}{\sqrt{\frac{p_1(1-p_1)}{n_1} + \frac{p_2(1-p_2)}{n_2}}} > z_{\alpha/2}$

**Contraste unilateral**
- Hipótesis nula $H_0: p_1 \leq p_2$
- Hipótesis alternativa $H_a: p_1 > p_2$

Se acepta $H_0$ si el estadístico $z = \frac{p_1 - p_2}{\sqrt{\frac{p_1(1-p_1)}{n_1} + \frac{p_2(1-p_2)}{n_2}}} \leq z_{\alpha}$

Se rechaza $H_0$ si el estadístico $z = \frac{p_1 - p_2}{\sqrt{\frac{p_1(1-p_1)}{n_1} + \frac{p_2(1-p_2)}{n_2}}} > z_{\alpha}$
